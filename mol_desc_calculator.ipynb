```python
from rdkit import Chem
from rdkit.Chem import Descriptors
import pandas as pd
```

```python
df = pd.read_csv(r'/home/lero/idrive/cmac/DDMAP/Stability_prediction_models/Smiles_calculator/api_descriptors.csv')
df
```

```python
def getMolDescriptors(mol, missingVal=None):
    results = {}
    for nm, fn in Descriptors._descList:
        try:
            val = fn(mol)
        except:
            # print the error message:
            import traceback
            traceback.print_exc()
            # and set the descriptor value to whatever missingVal is
            val = missingVal
        results[nm] = val
    return results
```

```python
descriptor_results = []

for index, row in df.iterrows():
    smiles = row['SMILES']
    try:
        mol = Chem.MolFromSmiles(smiles)
        if mol:
            descriptors = getMolDescriptors(mol)
            descriptor_results.append(descriptors)
        else:
            descriptor_results.append(None)
    except Exception as e:
        print(f"Error processing SMILES '{smiles}': {e}")

# Convert the list of dictionaries into a DataFrame
descriptor_df = pd.DataFrame(descriptor_results)

# Concatenate the original DataFrame with the descriptor DataFrame
final_df = pd.concat([df, descriptor_df], axis=1)
final_df = final_df.drop(['SMILES'], axis=1)

final_df.to_csv(r'/home/lero/idrive/cmac/DDMAP/Stability_prediction_models/Smiles_calculator/api_descriptors.csv', index=False)
```
