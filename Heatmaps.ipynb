```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable
from matplotlib.colors import TwoSlopeNorm
import os
```

```python
df = pd.read_csv(r"/home/lero/idrive/cmac/DDMAP/Stability studies/Stability_dataset_August_update.csv", na_values='nan')
df['Average days stable'] = (df['Average days stable'])/24

grouped_data = df.groupby(['API', 'condition'])
```

```python
norm = TwoSlopeNorm(vmin=0, vcenter=45, vmax=90)

for condition, condition_group in df.groupby('condition'):
    apis = condition_group['API'].unique
    
    figure, axes = plt.subplots(4,8, figsize=(20,15), dpi=700, sharey=True)
    axes=axes.flatten()
    cbar_ax = figure.add_axes([0.91, 0.3, 0.03, 0.4,])

    for i, (api, group) in enumerate(condition_group.groupby('API')):
        if i>=32:
            break
        ax=axes[i]
        heatmap_data = group.pivot_table(
        index='Drug loading (wt%)', 
        columns='Polymer',
        values='Average days stable',
        aggfunc='mean'  # or 'median', 'max' if more appropriate
    )
        heatmap_data=heatmap_data.sort_index(ascending=False)
        #move pure to last column
        if 'Pure' in heatmap_data.columns:
            cols = [col for col in heatmap_data.columns if col != 'Pure'] + ['Pure']
            heatmap_data = heatmap_data[cols]
            sns.heatmap(heatmap_data, ax=ax, cbar=(i==0), norm=norm, cbar_ax=None if i else cbar_ax,  cbar_kws={'label': 'Number of days stable'})
            ax.set_title(f'{api}')
            ax.set_xlabel('Polymer')
            ax.set_ylabel('Drug loading (wt%)')
            
    
    save_dir = '/home/lero/idrive/cmac/DDMAP/Stability studies/Model_results/Plots'
    plot_filename = os.path.join(save_dir, f'Stability_results_heatmap_{condition}.png')
    plt.tight_layout(rect=[0, 0, 0.9, 0.95])
    plt.suptitle(f'Heatmaps of stability results by API at {condition}', fontsize=18)
    plt.savefig(plot_filename)
    plt.show()
```
