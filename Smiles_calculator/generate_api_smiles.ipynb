```python
import pubchempy as pcp
import pandas as pd
```

```python
df  = pd.read_csv(r'/home/lero/idrive/cmac/DDMAP/Stability_prediction_models/Smiles_calculator/apis.csv')

smiles = []

for index, row in df.iterrows():
    api_name = row['API']
    compounds = pcp.get_compounds(api_name, 'name')

    if compounds:
        smiles.append(compounds[0].connectivity_smiles)
    else:
        print(f'error retrieving {api_name}')
        smiles.append(None)

df['SMILES'] = smiles
print(df)
df.to_csv(r'/home/lero/idrive/cmac/DDMAP/Stability_prediction_models/Smiles_calculator/api_descriptors.csv', index=False)
    
```
